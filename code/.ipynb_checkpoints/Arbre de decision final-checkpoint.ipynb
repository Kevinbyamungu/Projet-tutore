```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, DecisionTreeRegressor, plot_tree
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score, r2_score
import matplotlib.pyplot as plt
```

```python
# Exemple : ton fichier CSV
data = pd.read_csv("C:/Users/Kevin Byamungu/OneDrive/Documents/GitHub/Projet-tutore/Palmares/fichier finale.csv",sep=';')

# Afficher les premières lignes
print(data.head())
```

```python
print(data.columns.tolist())
```

```python
# Colonnes catégorielles à encoder
cat_cols = ['Option','TypeEcole', 'ProvinceEduc']

# Création d’un label encoder pour chaque colonne
encoder = LabelEncoder()
for col in cat_cols:
    data[col] = encoder.fit_transform(data[col].astype(str))
```

```python
# Transformation du taux de réussite en classes
def categoriser_reussite(x):
    if x < 50:
        return 'Faible'
    elif x < 75:
        return 'Moyenne'
    else:
        return 'Élevée'

data['CategorieReussite'] = data['TauxReussTot'].apply(categoriser_reussite)

# Définition des variables
X = data[['PartTot', 'PartH', 'PartF', 'Option', 'TypeEcole', 'ProvinceEduc']]
y = data['CategorieReussite']

# Division train/test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
model_reussite = DecisionTreeClassifier(criterion='entropy', max_depth=6, random_state=0)
model_reussite.fit(X_train, y_train)
```

```python
y_pred = model_reussite.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nMatrice de confusion:\n", confusion_matrix(y_test, y_pred))
print("\nRapport de classification:\n", classification_report(y_test, y_pred))
```

```python
plt.figure(figsize=(18,10))
plot_tree(model_reussite, feature_names=X.columns, class_names=model_reussite.classes_, filled=True, rounded=True)
plt.show()
```

```python
X = data[[ 'Option', 'TypeEcole', 'ProvinceEduc']]
y = data['TauxReussTot']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model_facteurs = DecisionTreeRegressor(criterion='friedman_mse', max_depth=6, random_state=0)
model_facteurs.fit(X_train, y_train)
```

```python
y_pred = model_facteurs.predict(X_test)
print("R² Score:", r2_score(y_test, y_pred))
```

```python
importances = pd.Series(model_facteurs.feature_importances_, index=X.columns)
print("Importance des variables :\n", importances.sort_values(ascending=False))

importances.sort_values().plot(kind='barh', figsize=(8,5), title="Facteurs influençant la réussite")
plt.show()
```

```python
# Création d'une variable écart de réussite
data['EcartReussite'] = data['TauxReussH'] - data['TauxReussF']

X = data[['Option', 'TypeEcole', 'ProvinceEduc']]
y = data['EcartReussite']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model_ecart = DecisionTreeRegressor(criterion='friedman_mse', max_depth=6, random_state=0)
model_ecart.fit(X_train, y_train)
```

```python
print("R² Score:", r2_score(y_test, model_ecart.predict(X_test)))
```

```python
importances = pd.Series(model_ecart.feature_importances_, index=X.columns)
importances.sort_values().plot(kind='barh', figsize=(8,5), title="Facteurs influençant les écarts garçons/filles")
plt.show()
```

```python
import pandas as pd
import joblib
import matplotlib.pyplot as plt
import seaborn as sns

# === 1. Charger les données que tu veux analyser (ex. un fichier Excel exporté depuis Power BI)
df = pd.read_excel("C:\\Users\\Kevin\\Documents\\donnees_powerbi.xlsx")

# === 2. Préparer les données (mêmes transformations que lors de l’entraînement)
# Si tu avais encodé Option, TypeEcole, ProvinceEduc dans ton modèle :
cat_cols = ['Option', 'TypeEcole', 'ProvinceEduc']
df = pd.get_dummies(df, columns=cat_cols, drop_first=True)

# === 3. Charger le modèle entraîné
model = joblib.load("C:\\Users\\Kevin\\Documents\\model_arbre_decision.pkl")

# === 4. Prédire la réussite
df['Prediction_Reussite'] = model.predict(df)

# === 5. (Optionnel) Probabilité de réussite
df['Prob_Reussite'] = model.predict_proba(df)[:, 1]

# === 6. Visualiser la distribution des probabilités
plt.figure(figsize=(6,4))
sns.histplot(df['Prob_Reussite'], bins=10, kde=True, color='skyblue')
plt.title("Distribution des probabilités de réussite")
plt.xlabel("Probabilité de réussite")
plt.ylabel("Nombre d’écoles")
plt.show()

# === 7. Exporter les résultats pour Power BI
df.to_excel("C:\\Users\\Kevin\\Documents\\resultats_prediction.xlsx", index=False)
print("✅ Prédictions enregistrées pour Power BI.")
```
